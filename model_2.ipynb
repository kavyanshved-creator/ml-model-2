```python
import pandas as pd 
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("advertising_cost_vs_sales_500.csv")

df.head()
```

```python
x = df[["AdvertisingBudget"]]
y = df["Sales"]

x.head()
y.head()
```

```python
x_train,x_test,y_train,y_test = train_test_split(
    x,y,test_size = 0.2 , random_state = 42
 )
```

```python
Sales_management = LinearRegression()

Sales_management.fit(x_train,y_train)
```

```python
y_pred = Sales_management.predict(x_test)

print(y_pred)
```

```python
print("actual data ")
print(y_test)

print( "predict dagta ")
print( y_pred)
```

```python
new_result = pd.DataFrame({
    "AdvertisingBudget" : [456]
})


budget = Sales_management.predict(new_result)
print("new result",budget[0])
```

```python
from sklearn.metrics import r2_score

y_pred = Sales_management.predict(x_test)

score = r2_score(y_test, y_pred)

print("R² Score:", score)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test,y_pred)

print(mae)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

```python
import joblib 

joblib.dump(Sales_management,"sales.pkl")

print("file saved successfully ")
```
